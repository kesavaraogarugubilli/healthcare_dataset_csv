## $$Total Billing

```python
total_billing=Info.groupby("Medical Condition")["Billing Amount"].sum().sort_values(ascending=False)
print(total_billing)
```

```python
Patient_count=Info["Medical Condition"].value_counts()
```

## Billing Amount 💹

```python
Info.sort_values(by="Billing Amount",ascending=False)
```

## Total Billing By Medical Condition

```python
total_billing.plot(
    kind="barh",
    figsize=(10,5),
    edgecolor="black",
    linewidth=0.5
)
plt.title("Total Billing by Medicalcondition",fontsize=12,fontweight="bold")
plt.xlabel("Total Billing Amount($)")
plt.ylabel("Medical Condition")
plt.grid(axis="x",linewidth=0.5,linestyle="--")

ax=plt.gca()
ax.xaxis.set_major_formatter(
   ticker.FuncFormatter(lambda x,pos:f"${x*1e-6:.1f}M")
)
plt.xlim(0,total_billing.max()*1.1)
plt.show()         
```

## No.of Patient Count By Medical Condition  

```python
Patient_count.plot(
    kind="barh",
    figsize=(20,5),
    edgecolor="black",
    color="lightgreen",
    linewidth=0.5
)
plt.title("No.of Patient count by Medicalcondition",fontsize=12,fontweight="bold")
plt.xlabel("Patient's count",color="black")
plt.ylabel("Medical Condition")
plt.grid(axis="x",linewidth=0.5,linestyle="--")

plt.xlim(0,Patient_count.max()*1.1)
plt.show()         
```

## Age Analysis Of Patients

```python
bins = [0,18,30,45,60,100]
labels = ["0-18","19-30","31-45","46-60","60+"]

Info["Age group"] = pd.cut(
    Info["Age"] , bins=bins , labels=labels
)

Age_group=Info["Age group"].value_counts().sort_index()
```

## Age Difference Index

```python
Age_group.plot(
kind="barh",
figsize=(10,5),
color="lightblue",
edgecolor="black")

plt.title("Age Difference Index",fontsize=10,fontweight="bold")
plt.xlabel("No.of patient")
plt.ylabel("Age group")

plt.show()
```

```python
gender_bill = Info.groupby("Gender")["Billing Amount"].sum().sort_index(ascending=False)
print(gender_bill)
```

## ♾️Distribution of Bills by Gender

```python
gender_bill.plot(
    kind="pie",
    autopct = "%2.4f%%",
    figsize=(5,3),
    startangle = 90
)
plt.title("Distribution of Bills by Gender", fontweight = "bold")
plt.axis("equal")
plt.tight_layout()
plt.show()
```

## Admission Type Distribution

```python
admission = Info["Admission Type"].value_counts()

plt.figure(figsize=(10,4))

plt.pie(
    admission,
    labels=admission.index,
    autopct="%2.2f%%",
    pctdistance=0.6, 
    wedgeprops=dict(width=0.6),
)

plt.title("Admission Type Distribution" , fontweight = "bold")
plt.tight_layout()
plt.show()
```

## Histogram Graph Of (Distribution Of blood) 

```python
blood = Info["Blood Type"].value_counts()

plt.figure(figsize=(10,6))

sns.histplot(data=Info, x="Blood Type",shrink=0.3, edgecolor="black")

plt.title("Distribution of Blood Types" , fontweight="bold")
plt.xlabel("Blood Type",fontweight="bold")
plt.ylabel("Count",fontweight="bold")
plt.grid(axis="y" , linestyle="--" , linewidth=0.5 , color="black")
plt.show()


```

```python
Info["Date of Admission"] = pd.to_datetime(Info["Date of Admission"])
Info["Discharge Date"] = pd.to_datetime(Info["Discharge Date"])

Info["length of stay"] = (Info["Discharge Date"] - Info["Date of Admission"]).dt.days
```

## Patient's Length Of Stay Graph

```python
plt.figure(figsize=(8,5))
plt.hist(Info["length of stay"],
         bins=24,
         edgecolor="black",
         color=("lightblue"))
plt.title("Distribution of Patient length of stay" , fontsize=14 , fontweight="bold")
plt.xlabel("Number of Days" , fontsize=12)
plt.ylabel("Number of Patients" , fontsize=12)
plt.grid(axis="y" , linestyle ="--" , alpha=0.7 , linewidth=0.5 , color="black")
plt.tight_layout()
plt.show()
```

## $Patient's Billing Amount Analysis

```python
plt.figure(figsize=(8, 4))

plt.boxplot(
    Info["Billing Amount"],
    vert=False,
    patch_artist=True,
    boxprops=dict(facecolor="lightgreen", color="black"),
    medianprops=dict(color="red", linewidth=4),
    flierprops=dict(marker="o", markerfacecolor="orange", alpha=0.7)
)

plt.title("Distribution of Patient's Billing Amount", fontsize=18, fontweight="bold")
plt.xlabel("Billing Amount ($)", fontsize=12)

plt.yticks([])

plt.grid(axis="x", linestyle="--", alpha=1.0 , color = "darkred" , linewidth = 0.6
         )

plt.tight_layout()  
plt.show()
```

## hexbin Of (Length Of Stay ♾️ Billing Amount) 

```python
plt.figure(figsize=(9, 6))

hb = plt.hexbin(
    Info["length of stay"],
    Info["Billing Amount"],
    gridsize=20,           
    cmap="Blues",          
    mincnt=1               
)

cb = plt.colorbar(hb)
cb.set_label("Number of Patients", fontsize=11)

plt.title("Relationship Between Length of Stay and Billing Amount", fontsize=14, fontweight="bold")
plt.xlabel("Length of Stay (Days)", fontsize=12)
plt.ylabel("Billing Amount ($)", fontsize=12)

plt.tight_layout()
plt.show()
```

## Graph Of (Insurance Provider ♾️ Billing Amount)

```python
insurance = (
    Info.groupby("Insurance Provider")["Billing Amount"]
    .sum()
    .sort_values(ascending=True)
)

plt.figure(figsize=(10, 5))


insurance.plot(
    kind="barh",          
    edgecolor="black",
    color="#065634"       
)

plt.title("Total Billing Amount by Insurance Provider", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Total Billing Amount ($)", fontsize=12)
plt.ylabel("Insurance Provider", fontsize=12)

plt.grid(axis="x", linestyle="--", linewidth=0.5, alpha=0.7 , color = "black")

plt.tight_layout()
plt.show()
```

## Analyzing patient's Test Results

```python
plt.figure(figsize=(7, 5))

ax = sns.countplot(
    data=Info,
    x="Test Results",
    order=Info["Test Results"].value_counts().index,  
    palette="Blues",                                 
    edgecolor="black"
)
ax.bar_label(ax.containers[0], padding=3, fontsize=10, fontweight="bold")

plt.title("Distribution of Test Results", fontsize=14, fontweight="bold", pad=12)
plt.xlabel("Test Result Category", fontsize=12)
plt.ylabel("Number of Patients", fontsize=12)

plt.grid(axis="y", linestyle="--", linewidth=0.5, alpha=0.7)

plt.tight_layout()
plt.show()
```

## Length Of Stay Visualisation

```python
plt.figure(figsize=(9, 6))

order = (
    Info.groupby("Medical Condition")["length of stay"]
    .median()
    .sort_values(ascending=True)
    .index
)

sns.boxplot(
    data=Info,
    y="Medical Condition",   
    x="length of stay",       
    order=order,             
    palette="Blues",        
    boxprops=dict(edgecolor="black"),
    medianprops=dict(color="red", linewidth=2)
)

plt.title("Length of Stay Distribution by Medical Condition", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Length of Stay (Days)", fontsize=12)
plt.ylabel("Medical Condition", fontsize=12)

plt.grid(axis="x", linestyle="--", linewidth=0.5, alpha=0.7)

plt.tight_layout()
plt.show()
```

## prescribed Medicinal Analysis

```python
medicine = (
    Info["Medication"]
    .value_counts()
    .head(10)
    .sort_values(ascending=True)
)
plt.figure(figsize=(9, 5))

ax = medicine.plot(
    kind="barh",
    color="darkgreen",
    edgecolor="black"
)

ax.bar_label(
    ax.containers[0],
    padding=5,
    fontsize=10,
    fontweight="bold"
)
plt.title("Top 10 Prescribed Medications", fontsize=14, fontweight="bold", pad=20)
plt.xlabel("Number of Prescriptions", fontsize=12)
plt.ylabel("Medication", fontsize=12)

plt.grid(axis="x", linestyle="--", linewidth=0.5, alpha=0.7, color ="black")
plt.xlim(0,medicine.max() * 1.12
)

plt.tight_layout()
plt.show()
```
